```python
def converRange(sqft):
    new_sqft=sqft.split("-")
    len_sqft=len(new_sqft)
    if(len_sqft==2):
        
        avg=(float(new_sqft[0])+ float(new_sqft[1]))/2
        
        return avg
    else:
        try:
            return float(new_sqft[0])
        except:
            None
```

```python
df['total_sqft']=df['total_sqft'].apply(converRange)
```

```python
def remove_outlier_price_per_sqft(df):
    df_output=pd.DataFrame()
    for key , subdf in  df.groupby('location'):
        print(key)
        m=np.mean(subdf['price_per_sqft'])
        sd=np.std(subdf['price_per_sqft'])
        gen_df=subdf[(subdf['price_per_sqft']>(m-3*sd)) & (subdf['price_per_sqft']<(m+3*sd))]
        df_output = pd.concat([df_output, gen_df], ignore_index=True)
    return df_output
```

```python
df=remove_outlier_price_per_sqft(df)
```
